```python
import pandas as pd

movies = pd.read_csv("movies.csv")
genres = pd.read_csv("genres.csv")
movie_genres = pd.read_csv("movie_genres.csv")
movie_keywords = pd.read_csv("movie_keywords.csv")
cast = pd.read_csv("cast.csv")
crew = pd.read_csv("crew.csv")
```

```python
print("Shape:", movies.shape)

print("\nColumn Names:")
print(movies.columns.tolist())

print("\nData Types:")
print(movies.dtypes)

print("\nMissing Values:")
print(movies.isnull().sum())
```

```python
# 1. Duplicate rows remove
movies = movies.drop_duplicates()
genres = genres.drop_duplicates()
movie_genres = movie_genres.drop_duplicates()
movie_keywords = movie_keywords.drop_duplicates()
cast = cast.drop_duplicates()
crew = crew.drop_duplicates()

# 2. Empty rows and columns remove
movies = movies.dropna(how="all").dropna(axis=1, how="all")
genres = genres.dropna(how="all").dropna(axis=1, how="all")
movie_genres = movie_genres.dropna(how="all").dropna(axis=1, how="all")
movie_keywords = movie_keywords.dropna(how="all").dropna(axis=1, how="all")
cast = cast.dropna(how="all").dropna(axis=1, how="all")
crew = crew.dropna(how="all").dropna(axis=1, how="all")


print("Movies:", movies.shape)
print("Genres:", genres.shape)
print("Movie Genres:", movie_genres.shape)
print("Movie Keywords:", movie_keywords.shape)
print("Cast:", cast.shape)
print("Crew:", crew.shape)
```

```python
# Convert release_date to date format
if "release_date" in movies.columns:
    movies["release_date"] = pd.to_datetime(
        movies["release_date"],
        errors="coerce"
    )

# Convert numeric columns
numeric_columns = [
    "budget",
    "revenue",
    "runtime",
    "popularity",
    "vote_average",
    "vote_count"
]

for column in numeric_columns:
    if column in movies.columns:
        movies[column] = pd.to_numeric(
            movies[column],
            errors="coerce"
        )

print("Data types updated!")

print(movies.dtypes)
```

```python
# Check missing values in all 6 files

dataframes = {
    "movies": movies,
    "genres": genres,
    "movie_genres": movie_genres,
    "movie_keywords": movie_keywords,
    "cast": cast,
    "crew": crew
}

for name, df in dataframes.items():
    print("\n", name)
    print(df.isnull().sum())
```

```python
# Save all 6 cleaned CSV files

movies.to_csv("movies.csv", index=False)
genres.to_csv("genres.csv", index=False)
movie_genres.to_csv("movie_genres.csv", index=False)
movie_keywords.to_csv("movie_keywords.csv", index=False)
cast.to_csv("cast.csv", index=False)
crew.to_csv("crew.csv", index=False)
```
